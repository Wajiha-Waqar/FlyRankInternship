## 3. Success metric

# Success Metric

The primary success metric is Precision@50.

Precision@50 measures how many of the top 50 recommended pages are actually positive according to the chosen label.

This metric matches the real decision process because content teams usually review only a limited number of pages. A ranked list is therefore more useful than overall accuracy.

If Precision@50 improves, reviewers spend more time on pages that genuinely deserve attention.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can identify obvious cases, such as pages with high impressions and long periods without updates. However, content performance depends on many interacting signals, including visibility, CTR, engagement, position, freshness, and traffic trends.

These relationships are difficult to capture using a small number of hand-written rules.

Machine learning can learn patterns across many signals simultaneously and assign different importance to different factors. This can produce a better ranked review queue than a single fixed rule.

The goal is not to replace human judgment. The goal is to help reviewers focus their limited time on the pages most likely to deserve attention.
