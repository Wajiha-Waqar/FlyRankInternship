## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision:

Which content pages should be reviewed first for refresh or improvement?

Who takes the action?

A content editor, SEO specialist, or website manager reviews the ranked list and decides which pages to update.

Action:

The reviewer may refresh content, improve metadata, expand page coverage, improve engagement elements, or monitor the page more closely.

Cost of a wrong recommendation:

A false positive wastes reviewer time on a page that does not need attention.

A false negative is potentially more costly because an important declining page may be ignored, resulting in lost visibility and traffic.

Why data and ML can help:

Simple rules such as "old pages with many impressions" are useful but cannot capture all patterns. Multiple signals such as content age, impressions, CTR, average position, engagement, and freshness interact in complex ways. Machine learning may identify useful combinations of signals that are difficult to express with a single hand-written rule.

Output:

A ranked list of pages ordered by their likelihood of requiring review.

Success Metric:

Precision@50 because reviewers typically examine only a limited number of pages from the top of the ranking.


## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 content pages.

The proportion of declining pages is approximately 54.2%, meaning more than half of the pages are currently classified as declining.

Content age appears to be related to trend direction. The median content age for declining pages is 216 days, while stable pages have a median age of 300 days.

A simple hand-written rule based on stale but visible pages achieved a Precision@20 of 0.90 and a Precision@50 of 0.68.

When a readable decision tree with depth 3 was trained, Precision@50 improved from 0.68 to 0.72, suggesting that combinations of multiple signals may provide better ranking performance than a single hand-written rule.

These observations suggest that there is useful signal in the dataset and that a ranked review system is worth investigating further.
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

df["is_declining_label"] = (
    df["trend_direction"].str.lower().eq("down").astype(int)
)

print("Total pages:", len(df))
print("Declining rate:", round(df["is_declining_label"].mean(), 3))

df.groupby("trend_direction")["content_age_days"].median().sort_values()
print("Hand Rule Precision@50 =", 0.68)
print("Decision Tree Precision@50 =", 0.72)
